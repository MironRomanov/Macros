```python
# Cell 2 — Install and start Ollama server
!curl -fsSL https://ollama.com/install.sh | sh

import subprocess
import os
import time

with open('ollama_server.log', 'w') as f:
    process = subprocess.Popen(
        ['ollama', 'serve'],
        stdout=f,
        stderr=f,
        preexec_fn=os.setsid
    )

# Give the server a moment to start up
time.sleep(3)
print('Ollama server started.')
```

```python
# Cell 3 — Pull the model (change model name here if needed)
MODEL = 'llama2'  # e.g. 'mistral', 'gemma', 'phi3'
!ollama pull {MODEL}
print(f'Model "{MODEL}" ready.')
```

```python
# Cell 4 — Interactive chatbot loop
# Type your message and press Enter. Type 'exit' or 'quit' to stop.

from langchain_community.llms import Ollama
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain

llm = Ollama(model=MODEL)

memory = ConversationBufferMemory()
chat = ConversationChain(llm=llm, memory=memory, verbose=False)

print('=' * 50)
print(f'Chatbot ready (model: {MODEL})')
print("Type 'exit' or 'quit' to stop.")
print('=' * 50)

while True:
    try:
        user_input = input('\nYou: ').strip()
    except (EOFError, KeyboardInterrupt):
        print('\nGoodbye!')
        break

    if not user_input:
        continue

    if user_input.lower() in ('exit', 'quit'):
        print('Goodbye!')
        break

    response = chat.predict(input=user_input)
    print(f'\nAI: {response}')
```
